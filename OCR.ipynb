```python
import pandas as pd
import os

BASE_DIR = r"C:\Users\LOQ\Downloads\MeterReading\MeterReading"
IMAGE_DIR = os.path.join(BASE_DIR, "images")
LABEL_FILE = os.path.join(BASE_DIR, "labels.csv")

df = pd.read_csv(LABEL_FILE)

print(df.head())
print("\nTotal images:", len(df))
```

```python
print("Missing image names:")
print(df["image"].isnull().sum())

print("\nMissing readings:")
print(df["reading"].isnull().sum())
```

```python
missing_images = []

for image_name in df["image"]:
    image_path = os.path.join(IMAGE_DIR, image_name)

    if not os.path.exists(image_path):
        missing_images.append(image_name)

print("Missing images:", len(missing_images))

if missing_images:
    print(missing_images)
```

```python
import matplotlib.pyplot as plt
import cv2

plt.figure(figsize=(15, 10))

for i, (_, row) in enumerate(df.head(12).iterrows()):

    image_path = os.path.join(IMAGE_DIR, row["image"])

    image = cv2.imread(image_path)

    if image is None:
        continue

    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    plt.subplot(3, 4, i + 1)
    plt.imshow(image)
    plt.title(f"Actual: {row['reading']}")
    plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
from sklearn.model_selection import train_test_split

train_df, val_df = train_test_split(
    df,
    test_size=0.20,
    random_state=42
)

print("Training images:", len(train_df))
print("Validation images:", len(val_df))
```

```python
TRAIN_DIR = os.path.join(BASE_DIR, "ocr_dataset", "train")
VAL_DIR = os.path.join(BASE_DIR, "ocr_dataset", "val")

os.makedirs(TRAIN_DIR, exist_ok=True)
os.makedirs(VAL_DIR, exist_ok=True)

print("Dataset folders created.")
```

```python
import shutil

def copy_images(dataframe, destination):

    for _, row in dataframe.iterrows():

        source = os.path.join(
            IMAGE_DIR,
            row["image"]
        )

        target = os.path.join(
            destination,
            row["image"]
        )

        shutil.copy2(source, target)


copy_images(train_df, TRAIN_DIR)
copy_images(val_df, VAL_DIR)

print("Images copied successfully.")
```

```python
import cv2
import matplotlib.pyplot as plt

IMAGE_PATH = r"C:\Users\LOQ\Downloads\MeterReading\MeterReading\images\m3.jpg"

img = cv2.imread(IMAGE_PATH)

if img is None:
    raise FileNotFoundError("Image could not be loaded")

img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

print("Image size:", img.shape)

plt.figure(figsize=(16, 10))
plt.imshow(img_rgb)
plt.axis("on")
plt.grid()
plt.show()
```

```python
x1 = 100
y1 = 100

x2 = 500
y2 = 300

crop = img[y1:y2, x1:x2]

plt.figure(figsize=(12, 5))
plt.imshow(cv2.cvtColor(crop, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```

```python
large = cv2.resize(
    crop,
    None,
    fx=5,
    fy=5,
    interpolation=cv2.INTER_CUBIC
)

plt.figure(figsize=(15, 6))
plt.imshow(cv2.cvtColor(large, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```

```python
import pytesseract

config = (
    "--oem 3 "
    "--psm 7 "
    "-c tessedit_char_whitelist=0123456789"
)

text = pytesseract.image_to_string(
    large,
    config=config
)

print("OCR:", repr(text))
```

```python
gray = cv2.cvtColor(
    large,
    cv2.COLOR_BGR2GRAY
)

# Otsu threshold
_, otsu = cv2.threshold(
    gray,
    0,
    255,
    cv2.THRESH_BINARY + cv2.THRESH_OTSU
)

# Adaptive threshold
adaptive = cv2.adaptiveThreshold(
    gray,
    255,
    cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
    cv2.THRESH_BINARY,
    31,
    10
)

# Inverted Otsu
inverted = cv2.bitwise_not(otsu)
```

```python
plt.figure(figsize=(15, 10))

plt.subplot(2, 2, 1)
plt.imshow(
    cv2.cvtColor(large, cv2.COLOR_BGR2RGB)
)
plt.title("Original")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(gray, cmap="gray")
plt.title("Grayscale")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(otsu, cmap="gray")
plt.title("Otsu")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(adaptive, cmap="gray")
plt.title("Adaptive")
plt.axis("off")

plt.show()
```

```python
images = {
    "original": large,
    "gray": gray,
    "otsu": otsu,
    "adaptive": adaptive,
    "inverted": inverted
}

for name, test_image in images.items():

    text = pytesseract.image_to_string(
        test_image,
        config=config
    )

    cleaned = "".join(
        c for c in text
        if c.isdigit()
    )

    print(
        f"{name:10} -> "
        f"raw={repr(text.strip())} "
        f"reading={cleaned}"
    )
```

```python
data = pytesseract.image_to_data(
    large,
    config=config,
    output_type=pytesseract.Output.DICT
)

for i in range(len(data["text"])):

    text = data["text"][i].strip()

    if text:

        confidence = data["conf"][i]

        print(
            "Text:",
            text
        )

        print(
            "Confidence:",
            confidence
        )
```
